```python
from manim import *
```

```python
%%manim -p Scene
class Scene(Scene):
	def construct(self):
		num_plane = NumberPlane(background_line_style={"stroke_color": BLUE,
													   "stroke_opacity": 0.3})
		self.add(num_plane)
		txt = Text("Hello mr. Bond.")
		self.play(Write(txt, rate_func=rate_functions.ease_in_sine))
		self.wait(2)
		txt.to_edge(UP)
		self.play(Unwrite(txt))

		sqr = Square(side_length=3)
		tor = Torus(color=PINK, major_radius=2, minor_radius=0.5)
		self.play(Write(sqr))
		self.wait(2)
		self.play(Transform(sqr, tor))
		self.wait(5)
```
